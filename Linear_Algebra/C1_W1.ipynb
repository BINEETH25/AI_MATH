# Advantages of using NumPy arrays:


Arrays are one of the core data structures of the NumPy library, essential for organizing the data.

```python
one_dimensional_arr = np.array([1, 2, 3, 4, 5])
print(one_dimensional_arr)
```

# np.arange

np.arange() :: This function will return an array of evenly spaced values within a given interval.
default starting integer -> 0.

```python
a = np.arange(3)
print(a)
b = np.arange(0,20,2)
print(b)
c = np.arange(-2, 3, 0.5)
print(c)
```

# np.linspace()

np.linspace() :: is a function that can create an array within evenly spaced values in given interval. (starting, ending, space_value)

Default Data Type : float type

```python
lin_spaced_arr = np.linspace(0, 10, 5)
print(lin_spaced_arr)
```
