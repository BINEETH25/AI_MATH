#1. System of Linear Equations.

:: A Linear system is <u><i><b>singular</b></i></u> if it has no unique solution, and otherwise, it is said to be <u><i><b>non-singular</b></i></u>.

:: A System of linear equations (or linear system) is a collection of one or more linear equations involving the same variables.


```python
import numpy as np
```

```python
A = np.array([
    [-1, 3],
    [3, 2]
    ], dtype= np.dtype(float))

B = np.array([7, 1], dtype = np.dtype(float))

print("Matrix A:")
print(A)
print("\nMatrix B:\n", B)
```

```python
print(f"Shape of A: {A.shape}")
print(f"Shape of B: {B.shape}")
```

```python
x = np.linalg.solve(A, B)
print(f"Solution x:\n{x}")
# np.linalg.solve() is a function in the NumPy library that solves a system of linear equations represented in matrix form. It takes two arguments: the coefficient matrix (A) and the constant matrix (B). The function returns the solution vector (x) that satisfies the equation Ax = B.
```

```python
d = np.linalg.det(A)
print(f"Determinant of A: {d:.2f}")
# The determinant of a matrix is a scalar value that can be computed from the elements of a square matrix. It provides important information about the matrix, such as whether it is invertible (non-zero determinant) or singular (zero determinant). In this case, the determinant of matrix A is calculated using the np.linalg.det() function from the NumPy library.
```

```python
A_system = np.hstack((A, B.reshape((2, 1))))
print(f"Augmented matrix [A|B]:\n{A_system}")
print("\n")
print(A_system[1])
```

```python
A_2 = np.array([
    [-1, 3],
    [3, -9]
    ], dtype=float)
B_2 = np.array([7, 1], dtype=float)

d_2 = np.linalg.det(A_2)
print(f"Determinant of A_2: {d_2:.2f}")
```

```python
try: 
    x_2 = np.linalg.solve(A_2, B_2)
except np.linalg.LinAlgError as err:
    print(err)
```

```python
A_2_system = np.hstack((A_2, B_2.reshape((2, 1))))
print(f"Augmented matrix [A_2|B_2]:\n{A_2_system}")
```
